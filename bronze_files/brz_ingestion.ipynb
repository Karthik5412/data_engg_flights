```python
dbutils.widgets.text('src',"")
table = dbutils.widgets.get('src')
```

```python
df = spark.readStream.format('cloudFiles')\
        .option('cloudFiles.format',"csv")\
        .option('cloudFiles.schemaLocation',f"/Volumes/flights/bronze/bronze_volumes/{table}/checkpoint/")\
        .option('cloudFiles.schemaEvolutionMode',"rescue")\
        .load(f"/Volumes/flights/source/data/{table}/")
```

```python
df.writeStream.format('delta')\
    .outputMode("append")\
    .trigger(once=True)\
    .option("checkpointLocation",f"/Volumes/flights/bronze/bronze_volumes/{table}/checkpoint")\
    .option("path",f"/Volumes/flights/bronze/bronze_volumes/{table}/data")\
    .start()
```
